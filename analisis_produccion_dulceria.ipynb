```python
# PROBLEMA: renombrar las columnas de los vencimientos
a = a.rename(columns={'DIA.1': 'DIA VENCIMIENTO',
                      'MES.1': 'MES VENCIMIENTO',
                      'AÑO.1': 'AÑO VENCIMIENTO'})
cols_venc = ['DIA VENCIMIENTO', 'MES VENCIMIENTO', 'AÑO VENCIMIENTO']

sin_venc = (a['PRESENTACION'] == 'PRUEBA') & a['FECHA VENCIMIENTO'].isna()
a.loc[sin_venc, cols_venc] = 0
a[cols_venc] = a[cols_venc].astype(int)

print('Pruebas completadas con 0:', sin_venc.sum())
print('Vacíos en DIA/MES/AÑO VENCIMIENTO:', a[cols_venc].isnull().sum().sum())
```

```python
#PROBLEMA: Hay columnas que tienen un espacio extra al final y genera problemas al intentar llamarlas despues.
a.columns = a.columns.str.strip()
a.columns = a.columns.str.replace(r'\s+', ' ', regex=True)
a.columns.tolist()
```

```python
#PROBLEMA: la columna KILOGRAMOS SOBRANTES tiene 95% nulos. Para tomar una decision, verificar si el mínimo es mayor a 0
#es una señal de que cuando no hay sobrante, la celda queda vacía en vez de cargar un 0.
a['KILOGRAMOS SOBRANTES'].describe()
```

```python
#El mínimo es 1.3, nunca 0. Ningún lote tiene registrado "0 kg sobrantes" explícitamente, cuando hay sobrante, se carga un valor positivo
#(desde 1.3 hasta 240 kg), y cuando no hay sobrante, la celda simplemente queda vacía.
a['KILOGRAMOS SOBRANTES'] = a['KILOGRAMOS SOBRANTES'].fillna(0)
a['KILOGRAMOS SOBRANTES'].describe()
```

Se sospecha un error de tipeo (posiblemente 24.0 kg con un cero de más), pero no se modifica el dato sin confirmación de la fábrica.
```python
#PROBLEMA: celdas vacias en columna "REPROCESO"
a['REPROCESO EN KG'] = a['REPROCESO EN KG'].fillna(0)
a['REPROCESO EN KG'].describe()
```

Los 708 valores vacíos de REPROCESO EN KG se completaron con 0, asumiendo que un lote sin reproceso registrado significa que no hubo reproceso (decisión basada en conocimiento del proceso de planta, no inferida únicamente de los datos).

```python
#PROBLEMA: celdas vacias
#La verificación correcta antes de decidir qué hacer con los nulos de LOTE. La lógica es: si la columna de cantidad usada
#(AZUCAR, SUERO, BICARBONATO, L.P.D) tiene un valor mayor a 0, significa que sí se usó el insumo y ahí un LOTE vacío sería un dato de
#trazabilidad genuinamente faltante (alguien no cargó de qué lote venía). Pero si la cantidad usada es 0 (o nula), tiene sentido que el lote
#también esté vacío, porque no había nada que trazar.
# AZUCAR
print("AZUCAR usada > 0 pero sin LOTE:",
      ((a['AZUCAR'] > 0) & (a['LOTE AZUCAR'].isnull())).sum())

# SUERO
print("SUERO usado > 0 pero sin LOTE:",
      ((a['SUERO'] > 0) & (a['LOTE SUERO'].isnull())).sum())

# BICARBONATO
print("BICARBONATO usado > 0 pero sin LOTE:",
      ((a['BICARBONATO'] > 0) & (a['LOTE BICARBONATO'].isnull())).sum())

# L.P.D.
print("L.P.D. usado > 0 pero sin LOTE:",
      ((a['L.P.D'] > 0) & (a['LOTE L.P.D.'].isnull())).sum())

# REPROCESO (ya con los nulos rellenados en 0, así que el chequeo es directo)
print("REPROCESO > 0 pero sin LOTE:",
      ((a['REPROCESO EN KG'] > 0) & (a['LOTE REPROSESO'].isnull())).sum())
```

```python
#Azúcar, Suero, Bicarbonato: 0 casos problemáticos.
#Para estos tres insumos, el patrón es perfectamente coherente: cada vez que se usó el insumo, el lote quedó registrado. Los nulos que
#tenías en esas columnas LOTE corresponden exactamente a los casos donde no se usó el insumo.
a['LOTE AZUCAR'] = a['LOTE AZUCAR'].fillna('No aplica')
a['LOTE SUERO'] = a['LOTE SUERO'].fillna('No aplica')
a['LOTE BICARBONATO'] = a['LOTE BICARBONATO'].fillna('No aplica')
#L.P.D.: 4 casos sin lote registrado; Reproceso: 366 casos sin lote registrado
a['LOTE L.P.D.'] = a['LOTE L.P.D.'].fillna('Sin registrar')
a['LOTE REPROSESO'] = a['LOTE REPROSESO'].fillna('Sin registrar')
```

```python
#PROBLEMA: vacios en el apartado presentaciones
columnas_pres = [c for c in a.columns if c.startswith('PRES ')]
columnas_cero = columnas_pres + ['KILOGRAMOS SOBRANTES']

a[columnas_cero] = a[columnas_cero].fillna(0)

a[columnas_cero].isnull().sum().sum()
```

```python
import re, numbers

# 1) Columnas PRES y sobrantes: todo a número, vacíos en 0
columnas_pres = [c for c in a.columns if c.startswith('PRES ')]
a[columnas_pres] = a[columnas_pres].apply(pd.to_numeric, errors='coerce').fillna(0)
a['KILOGRAMOS SOBRANTES'] = a['KILOGRAMOS SOBRANTES'].fillna(0)
a['REPROCESO EN KG'] = pd.to_numeric(a['REPROCESO EN KG'], errors='coerce').fillna(0)

# 2) Etiquetas de lote
for c in ['LOTE AZUCAR', 'LOTE SUERO', 'LOTE BICARBONATO']:
    a[c] = a[c].fillna('No aplica').replace('nan', 'No aplica')

a['LOTE REPROSESO'] = a['LOTE REPROSESO'].fillna('Sin registrar')
a['LOTE REPROSESO'] = a['LOTE REPROSESO'].replace(r'^\s*$', 'Sin registrar', regex=True)
a['LOTE L.P.D.'] = a['LOTE L.P.D.'].fillna('Sin registrar')

a.loc[(a['LOTE REPROSESO'] == 'Sin registrar') & (a['REPROCESO EN KG'] == 0), 'LOTE REPROSESO'] = 'No aplica'
a.loc[(a['LOTE L.P.D.'] == 'Sin registrar') & (a['L.P.D'] == 0), 'LOTE L.P.D.'] = 'No aplica'

# 3) Columnas LOTE: todo a texto, con prefijo "Lote" en lo que tiene números
def lote_a_texto(x):
    if isinstance(x, numbers.Number) and not pd.isna(x):
        return f'Lote {int(x)}' if float(x).is_integer() else f'Lote {x}'
    if isinstance(x, str) and re.search(r'\d', x) and not x.startswith('Lote '):
        return f'Lote {x}'
    return x

columnas_lote = [c for c in a.columns if c.startswith('LOTE ')]
for c in columnas_lote:
    a[c] = a[c].map(lote_a_texto).astype(str)

# Chequeos
etiquetas = {'No aplica', 'Sin registrar'}
print('PRES, vacíos:', a[columnas_pres].isnull().sum().sum())
print('Sin registrar -> REPROSESO:', (a['LOTE REPROSESO'] == 'Sin registrar').sum(),
      '| L.P.D.:', (a['LOTE L.P.D.'] == 'Sin registrar').sum())
for c in columnas_lote:
    sin_prefijo = sorted(v for v in a[c].unique() if v not in etiquetas and not v.startswith('Lote '))
    print(c, '| "nan":', (a[c] == 'nan').sum(), '| sin prefijo:', sin_prefijo)

# 4) Exportar (siempre al final)
ruta = '/content/registro_produccion_LIMPIO.xlsx'   # o la ruta de tu carpeta en Drive

with pd.ExcelWriter(ruta, engine='openpyxl', datetime_format='DD/MM/YYYY') as w:
    a.to_excel(w, sheet_name='Registro Produccion', index=False)
    ws = w.sheets['Registro Produccion']
    for nombre in ['FECHA ELABORACION', 'FECHA VENCIMIENTO']:
        letra = ws.cell(row=1, column=list(a.columns).index(nombre) + 1).column_letter
        ws.column_dimensions[letra].width = 14

from google.colab import files; files.download(ruta)   # descomentá para bajarlo a tu compu
```
